### A notebook remembers executed values
The runtime remembers variables created in cells you have executed. Editing a cell changes its text, but does not execute it. A previously calculated total does not automatically update when its inputs change.

Run the following two cells in order.

````python
order_score = 5
````

````python
order_total = order_score + 2
print("Total:", order_total)
````

**Check your output**

`Total: 7`

Now follow these steps:
1. Change `order_score = 5` to `order_score = 9`, but do not run that cell yet.
2. Run only the second cell. It still displays `7`, because the stored score is still `5`.
3. Run the first cell, then the second. It now displays `11`.
4. Restore `order_score = 5` and run both cells again. The output returns to `7`, matching the original worked example.

**Key idea:** changing an input does not automatically recalculate a previously stored result. Run the input cell and then the cells that depend on it.

If the runtime restarts, its variables are cleared. Your saved notebook remains, but you need to run its code again. Saved output can still be visible even when the current runtime has no variables. Use **Runtime → Run all** to execute from the top when checking your completed work.

## 7. Reuse Code with a Function

*Suggested time: 10 minutes*

You have calculated averages directly. If we want the same calculation for several pairs of scores, we can give it a name and reuse it. A **function** is a named piece of code that performs a task when called.

For today, follow and modify the supplied example. You do not need to design a function independently yet. Lab 5 will revisit functions before combining them with decisions.

### Define the Function

Read the next cell before running it:
- `def` begins a function definition.
- `calculate_average` is the name we choose.
- `score_1` and `score_2` are **parameters**: names for the values the function will receive.
- The colon `:` ends the definition header.
- The next line is indented by **four spaces**, showing that it belongs to the function. Keep this indentation.
- `return` calculates the expression and sends the result back to the code that called the function.

Run this definition cell first. No output is expected: defining the function makes it available; it does not yet run the calculation.

````python
def calculate_average(score_1, score_2):
    return (score_1 + score_2) / 2
````

### Call the Function

A **call** runs the function. In `calculate_average(6, 8)`, the supplied values `6` and `8` are called **arguments**. They are received by `score_1` and `score_2`, in that order.

**Before you run:** Predict the result. The returned value is stored in `function_average`, and `print()` displays it.

````python
function_average = calculate_average(8, 10)
print("Average:", function_average)
````

**Check your output**

```text
Average: 7.0
```

The flow is: **arguments 6 and 8 → parameters score_1 and score_2 → calculation → returned value 7.0 → store → display**.

`return` sends a value back; it does not display it. `print()` displays a value. The parameter names belong to the function call; you do not need to create them in a separate cell.
